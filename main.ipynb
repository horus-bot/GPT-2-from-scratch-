loading the data set 

```python
with open("input.txt", "r") as f:
    text = f.read()

```

fetching all the character

```python
chars = sorted(list(set(text)))
print(chars)
```

encoding and decoding functions

```python
numerize = {ch:i for i,ch in enumerate(chars)}
characterize ={i:ch for i,ch in enumerate(chars)}

```

```python
encode = lambda s : [numerize[c] for c in s ] 
decode = lambda s : "".join([characterize[c] for c in s])
```

encoding the data

```python
import torch
data = torch.tensor(encode(text),dtype=torch.long)
```

train and test split (90%)

```python
n = int(0.9 * len(data))
print(n)
train_data=data[:n]

val_data = data[n:]
```

chunking 

```python
torch.manual_seed(1337) #didnt understand that much but it creates same output , used fpr reproduction
batch_size = 4
block_size= 8
```

batching function

```python
def batch(split):
    data = train_data if split =="train"  else val_data
    ix = torch.randint(len(data)-block_size,(batch_size,))
    x = torch.stack([data[i:i+block_size] for i in ix])
    y = torch.stack([data[i+1:i+block_size+1] for i in ix])
    return x,y

a,b= batch("train")
print(a)
```
