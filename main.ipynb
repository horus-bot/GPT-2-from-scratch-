loading the data set 

```python
with open("input.txt", "r") as f:
    text = f.read()

```

fetching all the character

```python
chars = sorted(list(set(text)))
print(chars)
```

encoding and decoding functions

```python
numerize = {ch:i for i,ch in enumerate(chars)}
characterize ={i:ch for i,ch in enumerate(chars)}

```

```python
encode = lambda s : [numerize[c] for c in s ] 
decode = lambda s : "".join([characterize[c] for c in s])
```

encoding the data

```python
import torch
data = torch.tensor(encode(text),dtype=torch.long)
```

train and test split (90%)

```python
n = int(0.9 * len(data))
print(n)
test_data=data[:n]
print(test_data)
val_data = data[n:]
print(val_data)
```
